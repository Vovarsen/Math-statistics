```python
import numpy as np
import scipy.stats as st

thetta_true = 3
n = 100
beta = 0.95

Xn_ = st.pareto.rvs(thetta_true - 1, size=n)

thetta_OMP = 1 + n / np.sum(np.log(Xn_))

t1 = st.norm.ppf((1 - beta) / 2)
t2 = st.norm.ppf((1 + beta) / 2)

sigmma_thetta = (thetta_OMP - 1) / np.sqrt(n)
asymp_interv_thetta = [thetta_OMP - sigmma_thetta * t2, thetta_OMP - sigmma_thetta * t1]

median_OMP = 2**(1 / (thetta_OMP - 1))
sigmma_median = (median_OMP * np.log(2)) / ((thetta_OMP - 1) * np.sqrt(n))
asymp_interv_median = [median_OMP - sigmma_median * t2, median_OMP - sigmma_median * t1]


print("asymp_interv_thetta:", asymp_interv_thetta)
print("asymp_interv_median:", asymp_interv_median)
```

```python
N_boot_np = 1000
boot_thetta_np = []
for _ in range(N_boot_np):
    resample_np = np.random.choice(Xn_, size=n, replace=True)
    boot_thetta_np.append(1 + n / np.sum(np.log(resample_np)))

deltas_np = np.sort(np.array(boot_thetta_np) - thetta_OMP)
k1_np = int(N_boot_np * (1 - beta) / 2)
k2_np = int(N_boot_np * (1 + beta) / 2)
boot_interv_thetta_np = [thetta_OMP - deltas_np[k2_np-1], thetta_OMP - deltas_np[k1_np-1]]

N_boot_p = 50000
boot_thetta_p = []
for _ in range(N_boot_p):
    resample_p = st.pareto.rvs(thetta_OMP - 1, size=n)
    boot_thetta_p.append(1 + n / np.sum(np.log(resample_p)))

deltas_p = np.sort(np.array(boot_thetta_p) - thetta_OMP)
k1_p = int(N_boot_p * (1 - beta) / 2)
k2_p = int(N_boot_p * (1 + beta) / 2)
boot_interv_thetta_p = [thetta_OMP - deltas_p[k2_p-1], thetta_OMP - deltas_p[k1_p-1]]

print("boot_interv_thetta_np:", boot_interv_thetta_np)
print("boot_interv_thetta_p:", boot_interv_thetta_p)
```
