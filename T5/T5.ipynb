```python
n = 100
betta = 0.95

thetta = 50
Xn_ = np.random.uniform(low = thetta, high = 2* thetta, size = n)
print(Xn_)
print()

x_max = np.max(Xn_)
t1 = 1 + ((1-betta)/2) ** (1/n)
t2 = 1 + ((1+betta)/2) ** (1/n)
left_bord_exact = x_max / (t2)
right_bord_exact = x_max/(t1)
exact_interv = [left_bord_exact, right_bord_exact]
print("точный доверительный интервал:", exact_interv)
print()

thetta_OMM = (2/3) * np.mean(Xn_)
alpha_1 = np.mean(Xn_)
alpha_2 = np.mean(Xn_ **2)
u1_ = st.norm.ppf((1 - betta)/2)
u2_ = st.norm.ppf((1 + betta)/2)
coeff = ((2 * np.sqrt(alpha_2 - alpha_1**2)) / (3 * np.sqrt(n)))
asymp_interv = [thetta_OMM - coeff*u2_, thetta_OMM - coeff * u1_]
print("асимптотический доверительный интервал", asymp_interv)







```

```python
N_boot = 1000 
k1 = int(N_boot * (1 - betta) / 2) 
k2 = int(N_boot * (1 + betta) / 2) 

thetta_OMP = np.max(Xn_) / 2 
delta_OMM = []
delta_OMP = []

for i in range(N_boot):
    resample = np.random.choice(Xn_, size=n, replace=True)
    thetta_OMM_estim = (2/3) * np.mean(resample)
    thetta_OMP_estim = np.max(resample) / 2
    
    delta_OMM.append(thetta_OMM_estim - thetta_OMM)
    delta_OMP.append(thetta_OMP_estim - thetta_OMP)

delta_OMM_sorted = np.sort(delta_OMM)
delta_OMP_sorted = np.sort(delta_OMP)

boot_OMM = [thetta_OMM - delta_OMM_sorted[k2-1], thetta_OMM - delta_OMM_sorted[k1-1]]
boot_OMP = [thetta_OMP - delta_OMP_sorted[k2-1], thetta_OMP - delta_OMP_sorted[k1-1]]

print(f"Bootstrap OMM : {boot_OMM}")
print(f"Bootstrap OMP : {boot_OMP}")
```
